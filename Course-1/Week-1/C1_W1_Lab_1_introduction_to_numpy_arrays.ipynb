## 1.2 - Преимущества использования массивов NumPy ##

Массивы являются одной из основных структур данных библиотеки NumPy, необходимых для организации ваших данных. Вы можете думать о них как о сетке значений, все одного типа. Если вы раньше использовали списки Python, вы можете помнить, что они удобны, так как вы можете хранить разные типы данных. Однако списки Python ограничены в функциях и занимают больше места и времени для обработки, чем массивы NumPy.

NumPy предоставляет объект массива, который намного быстрее и компактнее, чем списки Python. Благодаря обширной интеграции API, библиотека предлагает множество встроенных функций, которые значительно упрощают вычисления всего несколькими строками кода. Это может быть огромным преимуществом при выполнении математических операций с большими наборами данных.

Объект массива в NumPy называется `ndarray`, что означает 'n-мерный массив'. Для начала вы будете использовать один из самых распространенных типов массивов: одномерный массив ('1-D'). Одномерный массив представляет собой стандартный список значений entirely в одном измерении. Помните, что в NumPy все элементы массива имеют одинаковый тип.

```python
one_dimensional_arr = np.array([10, 12])
print(one_dimensional_arr)
```

## 1.3 - Как создавать массивы NumPy ##

Есть несколько способов создать массив в NumPy. Вы можете создать одномерный массив, просто используя функцию `array()`, которая принимает список значений в качестве аргумента и возвращает одномерный массив.

```python
# Создайте и напечатайте массив NumPy 'a', содержащий элементы 1, 2, 3.
a = np.array([1, 2, 3])
print(a)
```

Другой способ реализовать массив — использовать `np.arange()`. Эта функция возвращает массив равномерно распределенных значений в заданном интервале. Чтобы узнать больше об аргументах, которые принимает эта функция, в Jupyter Notebook есть мощная функция, которая позволяет вам получить доступ к документации любой функции, просто нажав `shift+tab` на клавиатуре при нажатии на функцию. Попробуйте это для встроенной документации `np.arange()`.

```python
# Создайте массив с 3 целыми числами, начиная с целого числа по умолчанию 0.
b = np.arange(3)
print(b)
```

```python
# Создайте массив, который начинается с целого числа 1, заканчивается на 20, с шагом 3.
c = np.arange(1, 20, 3)
print(c)
```

Что если вы хотите создать массив с пятью равномерно распределенными значениями в интервале от 0 до 100? Как вы можете заметить, у вас есть 3 параметра, которые должна принимать функция. Один параметр — начальное число, в данном случае 0, конечное число 100 и количество элементов в массиве, в данном случае 5. NumPy имеет функцию, которая позволяет вам сделать именно это, используя `np.linspace()`.

```python
lin_spaced_arr = np.linspace(0, 100, 5)
print(lin_spaced_arr)
```

Вы заметили, что вывод функции представлен в форме значений с плавающей точкой (например, "... 25. 50. ...")? Причина в том, что тип по умолчанию для значений в функции NumPy `np.linspace` — это плавающая точка (`np.float64`). Вы можете легко указать тип данных с помощью `dtype`. Если вы обратитесь к встроенной документации функций, вы можете заметить, что большинство функций принимают необязательный параметр `dtype`. В дополнение к float, NumPy имеет несколько других типов данных, таких как `int` и `char`.

Чтобы изменить тип на целые числа, вам нужно установить dtype в `int`. Вы можете сделать это даже в предыдущих функциях. Не стесняйтесь попробовать и изменить ячейки для вывода желаемого типа данных.

```python
lin_spaced_arr_int = np.linspace(0, 100, 5, dtype=int)
print(lin_spaced_arr_int)
```

```python
c_int = np.arange(1, 20, 3, dtype=int)
print(c_int)
```

```python
b_float = np.arange(3, dtype=float)
print(b_float)
```

```python
char_arr = np.array(['Welcome to Math for ML!'])
print(char_arr)
print(char_arr.dtype) # Печатает тип данных массива
```

Вы заметили, что вывод типа данных массива `char_arr` — `<U23`?
Это означает, что строка (`'Welcome to Math for ML!'`) является 23-символьной (23) unicode строкой (`U`) на архитектуре с little-endian (`<`). Вы можете узнать больше о типах данных [здесь](https://docs.scipy.org/doc/NumPy-1.13.0/reference/arrays.dtypes.html).

## 1.4 - Подробнее о массивах NumPy ##

Одно из преимуществ использования NumPy заключается в том, что вы можете легко создавать массивы с помощью встроенных функций, таких как:
- `np.ones()` - Возвращает новый массив, устанавливая значения в единицы.
- `np.zeros()` - Возвращает новый массив, устанавливая значения в нули.
- `np.empty()` - Возвращает новый неинициализированный массив.
- `np.random.rand()` - Возвращает новый массив со случайными значениями.

```python
# Возвращает новый массив формы 3, заполненный единицами.
ones_arr = np.ones(3)
print(ones_arr)
```

```python
# Возвращает новый массив формы 3, заполненный нулями.
zeros_arr = np.zeros(3)
print(zeros_arr)
```

```python
# Возвращает новый массив формы 3, без инициализации записей.
empt_arr = np.empty(3)
print(empt_arr)
```

```python
# Возвращает новый массив формы 3 со случайными числами между 0 и 1.
rand_arr = np.random.rand(3)
print(rand_arr)
```

# 2 - Многомерные массивы #
С NumPy вы также можете создавать массивы более чем с одним измерением. В приведенных выше примерах вы имели дело с одномерными массивами, где вы можете получить доступ к их элементам, используя один индекс. Многомерный массив имеет более одного столбца. Думайте о многомерном массиве как о листе Excel, где каждая строка/столбец представляет измерение.

```python
# Создайте двумерный массив (2-D)
two_dim_arr = np.array([[1,2,3], [4,5,6]])
print(two_dim_arr)
```

Альтернативный способ создания многомерного массива — это изменение формы исходного одномерного массива. Используя `np.reshape()`, вы можете переставить элементы предыдущего массива в новую форму.

```python
# 1-D массив
one_dim_arr = np.array([1, 2, 3, 4, 5, 6])

# Многомерный массив с использованием reshape()
multi_dim_arr = np.reshape(
                one_dim_arr, # массив для изменения формы
               (2,3) # размерности нового массива
              )
# Напечатайте новый 2-D массив с двумя строками и тремя столбцами
print(multi_dim_arr)
```
